### 1. Get Three inputs from user (2 Inputs are float values (a and b) , 1 Input as String value as mentioned in given list(add, sub, mul, div)


```python

a = float(input('enter 1st number: '))
b = float(input('enter 2nd number: '))
print('first number is:',a)
print('second number is:',b)
c = input('Which action u wanna perform: \n(add, sub, mul, div)' )
if c == 'add':
  print('Addition of a and b is:', a + b)

elif c == 'sub':
  print('Diffrence of a and b is:', a - b)

elif c == 'mul':
  print('product of a and b is:',a * b)

elif c == 'div':
  if b == 0:
    print('Division is not possible due to divide by zero error')

  else:
    print('Divison of a and b is:', a/b)

else:
  print('Please enter valid input from list')
```

### 2. Create A python script to display below pattern using for loop


```python
a = int(input('enter number of lines u want:'))
for i in range(a,0,-1):
  for j in range(i,0,-1):
    print('*', end= '')
  print()
```

### 2. Create A python script to display below pattern using while loop


```python
a = int(input('enter number of lines u want:'))
while a >= 0:
  j = a
  a-=1
  while j >= 0:
    print('*', end= '')
    j-=1
  print()

```
